```python
journey = [
    "login",
    "view",
    "browse",
    "add_to_cart",
    "checkout"
]

text = " ".join(journey)
print(text)
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

embedding = model.encode(text)
```
